# Part 1 · Classes and objects

Every value in Python is an **object**, and every object belongs to a **class** (also called its *type*). You can think of a class as a **blueprint**, and of each object made from it as one particular thing built from that blueprint, called an **instance** of the class. `"cat"` and `"dog"` are two instances of the class `str`; `[1, 2]` and `[]` are two instances of the class `list`.

The class decides two things:

* what **data** each instance stores. For a list, that is its elements.
* what each instance can **do**: its **methods**. A method is a function that belongs to a class, and you call it with a dot after an object: `my_list.append(4)`.

A method works on the particular object it is called on. Every list has an `append` method, but `a.append(4)` changes `a` only:

```python
a = [1, 2, 3]
b = [10, 20]
a.append(4)               # changes a only
print(a, b)
print(type(a), type("hello"), type({}))
```

Writing your own class is useful whenever some data and the operations on it belong together. For example, a vocabulary needs both a store of words and counts (the data) and ways of adding words and looking up counts (the methods). Keeping them together in one class means the rest of your program only has to know *what* the methods do, not *how* they do it.

# Part 2 · Writing your own classes

Here is a class for a simple credit card. Each card stores its customer's name, a credit limit and a balance (how much has been spent). You can charge things to the card, as long as the balance doesn't go over the limit.

```python
class CreditCard:
    """A consumer credit card."""

    def __init__(self, customer, limit):
        """Create a new credit card. The balance starts at zero."""
        self._customer = customer
        self._limit = limit
        self._balance = 0

    def get_balance(self):
        """Return the current balance."""
        return self._balance

    def charge(self, price):
        """Charge price to the card if there is enough credit. Return True if it was charged."""
        if price + self._balance > self._limit:   # the charge would go over the limit
            return False
        else:
            self._balance += price
            return True
```

Let's go through it piece by piece:

* **`class CreditCard:`** starts the definition. Everything indented underneath belongs to the class. Class names are usually written in **CamelCase**: each word starts with a capital letter, with no underscores.
* Inside the class, each method is written with `def`, just like the functions in Week 4.
* **`__init__`** (with two underscores on either side) is a special method called the **constructor**. Writing `CreditCard("Ada Lovelace", 1000)` creates a new, empty object and then calls `__init__` to set it up, passing in `"Ada Lovelace"` and `1000`.
* **`self`** is the object that the method is working on. When you write `my_card.get_balance()`, Python passes `my_card` in as `self`. That is why each method is *defined* with an extra first parameter, `self`, which you leave out when you *call* it.
* **`self._balance = 0`** creates an **instance variable**: a piece of data that belongs to *this particular* object. Inside any method, `self._balance` means "this card's balance". A plain name such as `price` is just an ordinary local name (Week 4).
* A name starting with an underscore, such as `_balance`, is a signal to other programmers: *this is internal to the class, so don't use it from outside*. Use the class's methods instead, such as `get_balance()`. Keeping a class's internal details hidden like this is called **encapsulation**, and it means the inside of a class can be changed without breaking the code that uses it.
* The triple-quoted string at the start of the class and of each method is a **docstring**: documentation that `help()` displays.

Each object has its **own** instance variables. 🔮 **Predict** the printed values before running:

```python
my_card = CreditCard("Ada Lovelace", 1000)
your_card = CreditCard("Alan Turing", 500)

print(my_card.charge(300))
print(my_card.charge(800))
print(my_card.get_balance(), your_card.get_balance())
print(type(my_card))
help(CreditCard.charge)
```

> ⚠️ **Two common mistakes.** Forgetting the brackets, as in `my_card.get_balance`, gives you the method itself rather than calling it. And forgetting `self` in a method definition, as in `def get_balance():`, causes a `TypeError` when the method is called, because Python still passes the object in.

# Part 3 · Special methods

What happens if we print one of our objects?

```python
print(c)
```

Not very helpful: Python doesn't know how to show a `Corpus`, so it just gives its class and its memory address. We can fix this with a **special method**.

Python turns many operators and built-in functions into calls to methods with special names, which start and end with two underscores (you have already met one, `__init__`):

| You write | Python calls |
|---|---|
| `len(a)` | `a.__len__()` |
| `a[k]` | `a.__getitem__(k)` |
| `str(a)`, `print(a)` | `a.__str__()` |
| `a == b` | `a.__eq__(b)` |
| `a + b` | `a.__add__(b)` |

This is how `+` can add numbers but join strings: `int` and `str` each define their own `__add__`. If you define these methods in your own class, your objects work with `len`, `[]`, `print`, `==` and `+` just like built-in ones. There is also a bonus: if a class has `__len__` and `__getitem__`, Python can automatically **loop** over its objects with `for` and test membership with `in`.

# Part 4 · Inheritance

Classes are often related from general to specific: a cat *is a* mammal, and a mammal *is an* animal. **Inheritance** lets a new class, the **subclass** (or *child*), start from an existing class, the **superclass** (or *parent*). The subclass automatically gets all of the parent's methods. It can then:

* **override** a method: give it a new version that replaces the parent's
* **extend** the class: add brand-new methods

Here is a credit card that charges a **£5 fee** whenever a charge is refused. `class PenaltyCreditCard(CreditCard):` names the parent class in brackets. Inside, **`super()`** lets the child call the parent's version of a method, so the parent's code doesn't have to be written out again.

```python
class PenaltyCreditCard(CreditCard):
    """A credit card that charges a fee for refused charges."""

    def charge(self, price):
        success = super().charge(price)        # use CreditCard's charge method
        if not success:
            self._balance += 5                 # the fee
        return success

pc = PenaltyCreditCard("Ada", 100)             # __init__ is inherited from CreditCard
print(pc.charge(90), pc.charge(20), pc.charge(20))
print(pc.get_balance())                        # get_balance is inherited too
print(isinstance(pc, PenaltyCreditCard), isinstance(pc, CreditCard))
```

When you write `pc.charge(...)`, Python looks for `charge` first in the object's own class, `PenaltyCreditCard`, and only if it isn't there does it look in the parent class, `CreditCard`. So the child's version is used. For `pc.get_balance()`, the child has no version of its own, so the parent's is used. `isinstance(x, C)` asks whether `x` is an instance of the class `C`: a `PenaltyCreditCard` counts as a `CreditCard` too.
