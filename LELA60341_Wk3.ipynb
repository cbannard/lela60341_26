## Conditionals

```python
if first_condition:
    first_body
elif second_condition:
    second_body
else:
    third_body
```

Python tests the conditions in order and runs **exactly one** body: the first whose condition is true, or the `else` body if none is. You can have any number of `elif` clauses (including none), and the `else` is optional.

Because of truthiness (Week 2), a non-Boolean can be used as a condition. `if response:` is shorthand for `if response != '':`.

````python
temperature = 12          # 👉 try changing this value and re-running

if temperature < 0:
    description = "freezing"
elif temperature < 15:
    description = "cold"
elif temperature < 25:
    description = "mild"
else:
    description = "hot"

print(temperature, "is", description)
````

## Loops

### `while` loops

```python
while condition:
    body
```

Python tests the condition. If it is true, it runs the body and then tests again. As soon as the condition is false, the loop ends. (If it never becomes false, the loop runs for ever! Use the ⏹ stop button if that happens.)

This loop moves an index `j` along `data` until it finds an `'X'` or reaches the end. Because `and` short-circuits, `j < len(data)` is checked *before* `data[j]`, so we never use an index that is out of range:

````python
data = list("abcXdX")       # 👉 try list("abc") too
j = 0
while j < len(data) and data[j] != 'X':
    j += 1

if j < len(data):
    print("first X is at index", j)
else:
    print("no X found")
````

### `for` loops

A `for` loop goes through the elements of any **iterable** (list, tuple, str, set, dict, file, …) one at a time.

```python
for element in iterable:
    body          # the body can use `element`
```

On each pass, the loop variable is bound to the next element. It is an ordinary name: re-binding it inside the body has **no** effect on the original data.

````python
data = [3, 8, -2, 7]

total = 0
for val in data:
    total += val            # use the loop variable
print("sum:", total, "| built-in sum:", sum(data))

biggest = data[0]           # assumes a non-empty list
for val in data:
    if val > biggest:
        biggest = val
print("max:", biggest, "| built-in max:", max(data))

for ch in "hey":            # strings are iterable too
    print(ch, end=" ")
````

### Counting with a dictionary

A very common pattern in text processing is **counting with a dict**. `d.get(key, 0) + 1` works whether or not the key is already there.

Looping over `d.items()` gives you each `(key, value)` pair, which you can unpack straight into two names:

````python
counts = {}
for letter in "mississippi":
    counts[letter] = counts.get(letter, 0) + 1
print(counts)

for letter, n in counts.items():
    print(letter, "occurs", n, "times")
````

### Index-based `for` loops

Sometimes you need to know **where** an element is, not just what it is. `range(n)` produces the integers `0, 1, …, n-1`, which are exactly the valid indices of a sequence of length `n`. The standard idiom is:

```python
for j in range(len(data)):
    ... data[j] ...
```

`range` takes up to three arguments: `range(stop)`, `range(start, stop)` and `range(start, stop, step)`. As with slices, `stop` is **not included**, and `step` can be negative.

````python
data = [3, 8, -2, 11, 7]
big_index = 0
for j in range(len(data)):
    if data[j] > data[big_index]:
        big_index = j
print("largest is", data[big_index], "at index", big_index)

print(list(range(5)), list(range(2, 7)), list(range(0, 20, 5)), list(range(5, 0, -1)))
````

### `break` and `continue`

* `break` ends the **innermost** enclosing loop immediately.
* `continue` skips the rest of the current pass and moves on to the next one.

Use them sparingly. They are most useful when they save you from writing a complicated loop condition.

````python
data, target = [4, 9, 2, 7], 2
found = False
for item in data:
    if item == target:
        found = True
        break              # no need to look further
print("found?", found)

for val in [3, -1, 4, -5, 9]:
    if val < 0:
        continue           # skip negatives
    print(val, end=" ")
````

## Console input and output

### `print`

`print(a, b, c)` writes its arguments separated by spaces and followed by a newline. Arguments that aren't strings are converted with `str()`. `print()` on its own writes a blank line. Three **keyword parameters** change this:

* `sep=` sets the separator (default `' '`)
* `end=` sets what goes at the end (default `'\n'`)
* `file=` sends the output to a file (see Files, below)

````python
print("maroon", 5)
print("a", "b", "c", sep=":")
print("a", "b", "c", sep="")
print("no newline here", end=" --> ")
print("continued")
````

### `input`

`input(prompt)` shows the prompt, waits for the user to type a line and press Enter, and returns what they typed as a **string** (without the newline). To get a number, convert the string:

```python
year = int(input("In what year were you born? "))
```

If the user types several values on one line, `split()` breaks the string at whitespace into a list of strings:

```python
reply = input("Enter x and y, separated by spaces: ")
pieces = reply.split()
x = float(pieces[0])
y = float(pieces[1])
```

## Files

`open(filename, mode)` returns a **file object** that you use to read or write the file.

| Mode | Meaning |
|---|---|
| `'r'` | read (the default) |
| `'w'` | write: **replaces** any existing file |
| `'a'` | append to the end |

| Method | What it does |
|---|---|
| `fp.read()` | the whole (rest of the) file, as one string |
| `fp.readline()` | the next line (including its `'\n'`) |
| `fp.readlines()` | all remaining lines, as a list of strings |
| `for line in fp:` | loop over the remaining lines |
| `fp.write(s)` | write a string (**no** newline added) |
| `print(..., file=fp)` | print into the file |
| `fp.close()` | close the file and make sure everything is saved |

Most Python code opens files with a `with` block, which closes the file automatically at the end of the block:

```python
with open("sample.txt") as fp:
    contents = fp.read()
```

````python
# write a file (this replaces it if it already exists)
fp = open("ch1_sample.txt", "w")
fp.write("first line\n")
fp.write("second line\n")
print("third", "line", file=fp)
fp.close()

# read it back in different ways
with open("ch1_sample.txt") as fp:
    print(repr(fp.read()))          # repr() shows the hidden \n characters

with open("ch1_sample.txt") as fp:
    print(fp.readlines())

# append, then loop over the lines
with open("ch1_sample.txt", "a") as fp:
    fp.write("fourth line\n")

with open("ch1_sample.txt") as fp:
    for line in fp:
        print(line.rstrip())        # rstrip() removes the trailing newline
````
