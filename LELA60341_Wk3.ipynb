# Part 1 · Writing your own functions

You have already used many functions that come with Python: `len()`, `sorted()`, `print()`. In Week 1 you saw that a **function** is a named, reusable block of code: it takes inputs (**parameters**), does some work, and sends back a result with `return`.

You define your own function with `def`:

```python
def name(parameter1, parameter2):
    body                 # indented, like the body of a loop
    return result
```

Writing a function is worth it whenever you would otherwise copy and paste the same code: write it once, then **call** it as often as you like.

````python
def type_token_ratio(words):
    n_types = len(set(words))
    n_tokens = len(words)
    return n_types / n_tokens

text1 = "the cat sat on the mat".split()
text2 = "a dog and a cat and a bird".split()
print(type_token_ratio(text1))
print(type_token_ratio(text2))
````

A few things to notice:

* Defining a function doesn't run it. The body only runs when you **call** the function, with brackets: `type_token_ratio(text1)`.
* Inside the function, the parameter `words` refers to whatever you passed in.
* `return` ends the function and sends a value back. A function with no `return` gives back `None`.
* Names created inside a function (`n_types`, `n_tokens`) are **local**: they only exist while the function is running, and they don't clash with names outside it.

````python
def shout(text):
    loud = text.upper() + "!"
    return loud

print(shout("hello"))

def forgot_to_return(text):
    loud = text.upper() + "!"     # no return statement

print(forgot_to_return("hello"))  # None
# print(loud)                     # 👉 uncomment: NameError, because loud only existed inside shout()
````

### Default values and keyword arguments

A parameter can have a **default value**, which is used when the caller leaves it out. Callers can also name a parameter when they pass it (a **keyword argument**), which makes the call easier to read.

You have seen this already: `print(..., sep=":")` and `sorted(data, reverse=True)` both use keyword arguments.

````python
def ngrams(words, n=2):
    result = []
    for i in range(len(words) - n + 1):
        result.append(" ".join(words[i:i+n]))
    return result

words = "the cat sat on the mat".split()
print(ngrams(words))           # n is 2 by default
print(ngrams(words, 3))        # n is 3
print(ngrams(words, n=1))      # the same thing, with a keyword argument
````

### Passing a function to a function: `key=`

Functions are objects too, so you can pass one to another function. The most useful case is the `key=` argument of `sorted()`, `max()` and `min()`: Python calls your function on each element and sorts (or compares) by the result.

For a tiny function you only need once, you can write a **lambda**: `lambda w: w[-1]` means "a function that takes `w` and returns `w[-1]`".

````python
words = ["pear", "fig", "banana", "kiwi"]
print(sorted(words))                   # alphabetical
print(sorted(words, key=len))          # by length
print(max(words, key=len))             # the longest word

def last_letter(w):
    return w[-1]
print(sorted(words, key=last_letter))           # by last letter
print(sorted(words, key=lambda w: w[-1]))       # the same, with a lambda

counts = {"the": 5, "cat": 2, "sat": 3}
print(sorted(counts.items(), key=lambda pair: pair[1], reverse=True))   # by count, largest first
````

## `enumerate` and `zip`

* `enumerate(seq)` gives you each element **together with its index**, so you don't need `range(len(...))`.
* `zip(a, b)` walks along two sequences **side by side**, giving you one element from each at a time.

````python
words = ["the", "cat", "sat"]

for i in range(len(words)):          # the Week 2 way
    print(i, words[i])

for i, word in enumerate(words):     # the same, with enumerate
    print(i, word)

tags = ["DET", "NOUN", "VERB"]
for word, tag in zip(words, tags):
    print(word, "/", tag)
````

## Comprehensions

A **list comprehension** builds a new list from an old one in a single line. These two do the same thing:

```python
lengths = []                           lengths = [len(w) for w in words]
for w in words:
    lengths.append(len(w))
```

Read it as "**`len(w)`** for each `w` in `words`". You can add an `if` to keep only some elements:

```python
long_words = [w for w in words if len(w) > 3]
```

There are **dict comprehensions** too: `{w: len(w) for w in words}`.

## Conditional expressions

`x if condition else y` is a one-line `if/else` that gives a **value**:

```python
label = "positive" if stars >= 3 else "negative"
```

A related trick: `True` and `False` behave like `1` and `0` in arithmetic, so `int(stars >= 3)` gives 1 or 0, and `sum()` of a list of booleans counts how many are `True`.

````python
words = ["the", "old", "woman", "was", "asleep"]

print([len(w) for w in words])
print([w.upper() for w in words])
print([w for w in words if len(w) > 3])
print({w: len(w) for w in words})

stars = [5, 1, 4, 2]
print(["positive" if s >= 3 else "negative" for s in stars])
print([int(s >= 3) for s in stars])
print(sum([s >= 3 for s in stars]), "positive reviews")

numbers = ["1.5", "2.25", "-0.5"]          # numbers read from a file arrive as strings
print([float(x) for x in numbers])
````

## Unpacking and simultaneous assignment

You have already unpacked tuples in loops (`for word, count in counts.items()`). You can unpack any sequence into names, and you can assign several names at once:

* `a, b = b, a` swaps two values.
* `first, *rest = words` puts the first element in `first` and **the rest** in a list called `rest`.
* `zip(*pairs)` "unzips" a list of pairs back into two sequences: the `*` unpacks the list into separate arguments.

````python
a, b = 1, 2
a, b = b, a                      # swap
print(a, b)

first, *rest = ["the", "cat", "sat", "down"]
print(first, rest)

pairs = [("the", 5), ("sat", 3), ("cat", 2)]
words, counts = zip(*pairs)      # unzip into two tuples
print(words, counts)
````

## Set operations

Sets (Week 2) support the operations you know from maths. They are a quick way to compare vocabularies:

| Operation | Meaning |
|---|---|
| `a \| b` | **union**: in `a` or `b` (or both) |
| `a & b` | **intersection**: in both |
| `a - b` | **difference**: in `a` but not in `b` |

````python
a = {"the", "cat", "sat"}
b = {"the", "dog", "sat", "down"}
print(a | b)
print(a & b)
print(a - b)
print(set(range(10)) - {2, 3, 5, 7})     # the numbers 0-9 that are not in the second set
````

# Part 3 · Errors and exceptions

When something goes wrong, Python **raises an exception** and stops with an error message (a *traceback*). The last line of the traceback tells you the **type** of error, which is usually the best clue:

| Error | Typical cause |
|---|---|
| `NameError` | a name that hasn't been defined (often a typo, or a cell you haven't run) |
| `KeyError` | looking up a key that isn't in a dictionary |
| `IndexError` | an index past the end of a list |
| `ValueError` | the right type but a bad value, e.g. `int("five")` or `vocab.index("zebra")` |
| `TypeError` | the wrong type, e.g. `"3" + 4` |
| `ZeroDivisionError` | dividing by zero |
| `IndentationError` | the indentation of a block is wrong |

````python
counts = {"cat": 2}
experiments = ['counts["dog"]', '["a", "b"][5]', 'int("five")', '"3" + 4', '1 / 0']
for code in experiments:
    try:
        eval(code)
    except Exception as err:
        print(f"{code:16} -> {type(err).__name__}: {err}")
````

### Catching an exception: `try` / `except`

If you expect that something might fail, put it in a `try` block and say what to do instead in an `except` block. Name the error type you expect, so that other mistakes still show up:

```python
try:
    rating = int(text)
except ValueError:
    rating = None          # what to do if text isn't a whole number
```

Often you can avoid the error in the first place: `word in counts` before `counts[word]`, or `counts.get(word, 0)`. Both styles are fine.

You can also **raise** an exception yourself, to stop with a clear message when something is wrong:

```python
if len(words) == 0:
    raise ValueError("words must not be empty")
```

````python
def safe_int(text):
    try:
        return int(text)
    except ValueError:
        return None

print(safe_int("42"), safe_int("forty-two"))
````

## `random`

The `random` module makes **pseudo-random** numbers: they look random, but they come from a formula. If you set the **seed** first, you get the same "random" results every time, which makes experiments repeatable.

````python
import random

random.seed(1)                       # 👉 change the seed, or remove this line, and run again
print(random.random())               # a float between 0 and 1
print(random.randint(1, 6))          # a whole number from 1 to 6
words = ["the", "cat", "sat", "on", "the", "mat"]
print(random.choice(words))          # one element at random
print(random.sample(words, 3))       # 3 different positions, at random
random.shuffle(words)                # shuffle the list IN PLACE (it returns None)
print(words)
````

## `collections.Counter`

Counting things is so common that Python has a ready-made tool for it. A `Counter` is a dictionary that counts for you, and its `.most_common(n)` method gives the `n` most frequent items, largest first, as `(item, count)` pairs.

````python
from collections import Counter

words = "the cat sat on the mat and the dog sat on the log".split()
counts = Counter(words)
print(counts["the"], counts["zebra"])    # a missing key counts as 0, with no KeyError
print(counts.most_common(3))
````
