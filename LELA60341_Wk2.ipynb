## Numbers: `int` and `float`

* `int` holds whole numbers of **any size**: Python never overflows.
* `float` holds numbers with a decimal point: `98.6`, `2.0`, or scientific notation `6.022e23` (6.022 × 10²³).

**Converting between types** is something you will do frequently, especially when reading text from the keyboard or a file, which always arrives as a string:

| Call | Result |
|---|---|
| `int('137')` | `137` (an invalid string raises `ValueError`) |
| `int(3.99)` | `3` (**truncates** toward zero, it does not round) |
| `float('3.14')`, `float(2)` | `3.14`, `2.0` |
| `str(42)` | `'42'` |
| `round(3.14159, 2)` | `3.14` |

**Arithmetic operators**

| Operator | Meaning |
|---|---|
| `+` `-` `*` | addition, subtraction, multiplication |
| `/` | **true division**: always gives a `float` |
| `//` | **integer (floor) division** |
| `%` | **modulo**: the remainder |
| `**` | exponentiation |

`count += 1` is shorthand for `count = count + 1`. There are versions for the other operators too (`-=`, `*=`, …).

```python
print(2 ** 200)                                      # no overflow, just a big int
print(int("137"), int(3.99), float("3.14"), str(42) + "!")
print(27 / 4, 27 // 4, 27 % 4)
print(type(27 / 4), type(27 // 4), 8 / 2)            # / always gives a float

count = 0
count += 5
count += 1
print(count, round(3.14159, 2))
```

## `bool`, comparisons and logical operators

A `bool` is either `True` or `False`. You usually get one by **comparing** things:

| Operator | Meaning |
|---|---|
| `==` / `!=` | equal / not equal values |
| `<` `<=` `>` `>=` | ordering |
| `is` / `is not` | the very same object (Week 1) |

Strings are compared **lexicographically** (dictionary order, character by character) and **case-sensitively**: every capital letter comes before every lower-case letter.

Booleans are combined with `and`, `or` and `not`. `and` and `or` **short-circuit**: in `a and b`, if `a` is false, `b` is never evaluated.


```python
print(3 < 5, 3 == 3.0, 3 != 4)
print("apple" < "banana", "Zebra" < "apple", "abc" < "abd")

x = 7
print(x > 0 and x < 10, x < 0 or x > 5, not x > 5)
```

## Sequence types: `list`, `tuple` and `str`

A **sequence** is a collection in which **order matters**. Sequences are **zero-indexed**: a sequence of length *n* has indices `0` to `n − 1`. A `list` can hold any objects and is **mutable**; a `str` holds characters and is immutable.

### Working with lists

| Operation | Result |
|---|---|
| `data[i]`, `data[-1]` | element at index `i` / the last element |
| `len(data)` | number of elements |
| `x in data` | `True` if some element equals `x` |
| `data.index(x)`, `data.count(x)` | position of the first `x` / how many `x` there are |
| `data.append(x)` | add `x` at the **end** |
| `data.insert(i, x)` | add `x` **at index `i`** |
| `data[i] = x` | replace the element at index `i` |
| `data.pop()` | remove **and return** the last element |
| `data.remove(x)` | remove the **first** element equal to `x` |
| `sorted(data)` | a **new**, sorted list (Week 1) |

Methods that change a list in place (like `append`) return `None`, as we saw in Week 1.

```python
colours = ["red", "green", "blue"]
print(colours[0], colours[-1], len(colours), "green" in colours, colours.index("blue"))

colours.append("yellow")          # at the end
colours.insert(1, "orange")       # at index 1
colours[0] = "crimson"            # replace
print(colours)

last = colours.pop()              # returns what it removed
colours.remove("blue")            # by value
print("removed", last, "->", colours)
```

### Slicing and other sequence operators

`str`, `tuple` and `list` all support:

| Syntax | Meaning |
|---|---|
| `s[j]` | element at index `j` |
| `s[start:stop]` | slice with indices `start` up to, **not including**, `stop` |
| `s[start:stop:step]` | every `step`-th index from `start`, stopping before `stop` |
| `s + t` | concatenation |
| `k * s` | `s + s + ... + s` (`k` times) |
| `val in s` / `val not in s` | containment check |

* **Negative indices** count from the end: `s[-1]` is the last element and `s[-2]` the second to last.
* Slices are **half-open**: `data[3:8]` covers indices 3, 4, 5, 6 and 7. Leave out `start` or `stop` to mean "from the beginning" or "to the end". A negative `step` walks backwards.

```python
data = list(range(10, 20))        # [10, 11, ..., 19]
print(data)
print(data[0], data[-1], data[-2])
print(data[3:8])                  # indices 3..7
print(data[:3], data[7:])         # omitted start / stop
print(data[::2], data[::-1])      # every 2nd; reversed
print([0] * 4 + [1, 2])
print("amp" in "example", 15 in data)
```

### `tuple`

A tuple is an immutable list, written with parentheses: `(3, 4)`. A one-element tuple needs a trailing comma, `(17,)`, because `(17)` is just the number 17 in brackets.

**Accessing** a tuple works exactly like a list: `t[i]`, `t[-1]`, slices `t[i:j]`, `len(t)`, `x in t`, `t.index(x)` and `t.count(x)`. Tuples are also often **unpacked** into separate names: `name, age = ("Ada", 36)`.

Tuples are **immutable**, so there is no `append`, `insert`, `remove` or `pop`, and you can't write `t[i] = x` or `del t[i]`. To "add", "change" or "remove" an element, you build a **new** tuple, usually with `+` and slicing:

| Goal | Idiom | Note |
|---|---|---|
| add at the end | `t + (x,)` | the trailing comma makes `(x,)` a tuple |
| add at index `i` | `t[:i] + (x,) + t[i:]` | |
| change index `i` | `t[:i] + (x,) + t[i+1:]` | |
| remove index `i` | `t[:i] + t[i+1:]` | |
| many changes | `tmp = list(t)` … edit `tmp` … `t = tuple(tmp)` | convert to a list and back |

The original tuple never changes. Any other names that refer to it still see the old values.

**Comparing and sorting tuples.** Tuples (and lists) are compared element by element, so `(5, 'a') < (5, 'b') < (6, 'a')`. This means that sorting a list of `(count, word)` pairs sorts them by count first, then alphabetically.

```python
point = (3, 4, 5)
print(point[0], point[-1], point[:2], len(point), 4 in point, point.index(5))
x, y, z = point                       # unpacking
print(x, y, z)

bigger  = point + (6,)                # "add"
changed = point[:1] + (99,) + point[2:]   # "change" index 1
smaller = point[:1] + point[2:]       # "remove" index 1
print(point, bigger, changed, smaller)

pairs = [(3, "cat"), (5, "dog"), (1, "ant"), (5, "bee")]
print(sorted(pairs))                  # by count, then by word
print(sorted(pairs, reverse=True))    # largest count first
```

### `str`

A string is an immutable sequence of characters, written in single or double quotes: `'hello'` or `"Don't worry"`. A backslash starts an **escape character**: `\n` is a newline, `\t` is a tab, and `\'` is a quote.

| Operation | Result |
|---|---|
| `s[i]`, `s[i:j]`, `len(s)` | a character / a substring / the length |
| `sub in s` | is `sub` a **substring** of `s`? |
| `s.find(sub)` | index where `sub` first occurs, or **-1** if it doesn't |
| `s.count(sub)` | number of occurrences |
| `s.startswith(p)` | does `s` begin with `p`? |
| `s.lower()`, `s.upper()` | a lower- / upper-case copy |
| `s.replace(old, new)` | a copy with every `old` replaced by `new` (use `""` to remove it) |
| `s.strip()`, `s.rstrip()` | a copy without whitespace at both ends / the right-hand end |
| `s.split()`, `s.split(",")` | a **list** of the pieces between whitespace / commas |
| `"-".join(list_of_strings)` | the pieces glued together with `"-"` between them |
| `f"{name} is {age}"` | an f-string: values inserted into the text |

Strings are **immutable**: every method returns a **new** string. To keep a result, assign it: `s = s.lower()`.

```python
s = "Hello, world"
print(s[0], s[-5:], len(s), "world" in s, s.find("o"), s.count("l"), s.startswith("Hell"))
print(s.lower(), s.upper(), s.replace("world", "Python"), s.replace(",", ""))
print(repr("  lots of space  ".strip()))

words = "the quick brown fox".split()
print(words, "-".join(words))
print(f"{s} has {len(s)} characters")
print("col1\tcol2\nrow2")        # a tab and a newline
```

## `set`

A **set** is a mathematical set: **no duplicates** and **no order**. Membership tests (`x in s`) are very fast. Only **immutable** objects (`int`, `float`, `str`, `tuple`, …) can be elements.

`set(sequence)` is a quick way to find the **distinct** elements of a list or string.

```python
colours = {"red", "green", "blue", "red"}   # duplicate is dropped
print(colours, len(colours))
print("green" in colours)

words = "the cat sat on the mat".split()
print(len(words), "words but only", len(set(words)), "different words:", set(words))
print(type({}), type(set()))                 # {} is an empty DICT, not a set!
```

## The `dict` class

A **dictionary** (or *mapping*) links a set of distinct **keys** to associated **values**, for example student IDs to student records. It uses the same hash-table idea as a set, so keys must be hashable, and looking up a key is fast.

Ways to make one:

* Literal: `{'ga': 'Irish', 'de': 'German'}`. An empty `{}` is an empty dict.
* `dict(existing_mapping)` makes a new dict with the same associations.
* `dict(pairs)` builds a dict from a sequence of `(key, value)` pairs.

### Working with dictionaries

Dictionaries are **mutable** and are always accessed **by key**, never by position.

| Operation | Result |
|---|---|
| `d[key]` | the value for `key` (raises `KeyError` if it's missing) |
| `d.get(key, default)` | the value, or `default` if `key` is missing (no error) |
| `key in d`, `len(d)` | is `key` one of the keys? / number of pairs |
| `d.keys()`, `d.values()`, `d.items()` | the keys, the values and the `(key, value)` pairs |
| `d[key] = value` | add a new pair, or replace the value if `key` exists |
| `d[key] += 1` | update a value (the key must already exist) |
| `d.pop(key)` | remove the pair **and return its value** |

```python
capitals = {"France": "Paris", "Japan": "Tokyo"}
print(capitals["France"], capitals.get("Peru", "unknown"), "Japan" in capitals, len(capitals))
print(list(capitals.keys()), list(capitals.values()), list(capitals.items()))

capitals["Peru"] = "Lima"            # new key -> added
capitals["Japan"] = "Tokyo (東京)"    # existing key -> replaced
peru = capitals.pop("Peru")
print("popped", peru, "->", capitals)
```
