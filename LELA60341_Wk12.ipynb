# Part 5 · A layer with no weights: ReLU

In Week 8 you wrote a `Linear` layer with three methods: `forward`, `backward` and `update`. A layer doesn't have to have any weights. A **ReLU layer** just applies `relu` to every number it is given. It has nothing to learn, but it still has the same three methods, so that it can be used in the same way as any other layer.

Run the next cell first: it sets up the `Linear` layer from Week 8, with the helper functions `dot`, `mat_vec` and `transpose`.

````python
import math
import random
import matplotlib.pyplot as plt

def dot(a, b):
    # the dot product of two lists of the same length
    total = 0
    for i in range(len(a)):
        total += a[i] * b[i]
    return total

def mat_vec(matrix, vector):
    # the dot product of each row of the matrix with the vector
    return [dot(row, vector) for row in matrix]

def transpose(matrix):
    # the rows of the matrix turned into columns
    return [[row[j] for row in matrix] for j in range(len(matrix[0]))]

def sigmoid(z):
    if z >= 0:
        return 1 / (1 + math.exp(-z))
    else:
        return math.exp(z) / (1 + math.exp(z))      # the same value, without calculating math.exp(-z)

class Linear:
    """A layer that gives n_out scores for each example: one row of weights and one bias per output."""
    def __init__(self, n_in, n_out):
        self.weight = [[random.random() for j in range(n_in)] for c in range(n_out)]
        self.bias = [random.random() for c in range(n_out)]

    def forward(self, X):
        self.inputs = X                       # kept for the backward pass
        scores = []
        for x in X:
            z = mat_vec(self.weight, x)
            for c in range(len(z)):
                z[c] += self.bias[c]
            scores.append(z)
        return scores

    def backward(self, grad):
        n_out = len(self.weight)
        n_in = len(self.weight[0])
        self.weight_grad = [[0] * n_in for c in range(n_out)]
        self.bias_grad = [0] * n_out
        for k in range(len(self.inputs)):
            for c in range(n_out):
                for j in range(n_in):
                    self.weight_grad[c][j] += grad[k][c] * self.inputs[k][j]
                self.bias_grad[c] += grad[k][c]
        # the gradients for the inputs, for a layer that comes before this one
        return [mat_vec(transpose(self.weight), g) for g in grad]

    def update(self, lr):
        for c in range(len(self.weight)):
            for j in range(len(self.weight[c])):
                self.weight[c][j] -= lr * self.weight_grad[c][j]
            self.bias[c] -= lr * self.bias_grad[c]
````

## Step 1 · The forward pass

`forward(X)` takes a matrix (one row per example) and returns a matrix of the same shape with `relu` applied to every number. It also keeps `X` in `self.inputs` for the backward pass. `update(lr)` does nothing, because there are no weights to change: its body is `pass`, which is Python for "do nothing".

|---|---|---|
| example 0 | `[-1, 2, 0]` | `[0, 2, 0]` |
| example 1 | `[3, -4, 5]` | `[3, 0, 5]` |

## Step 2 · The backward pass and the update

The backward pass goes through the layers in **reverse order**, because it works from the output back to the input: each layer's `backward` is given the gradients from the layer after it, and returns the gradients for the layer before it. `reversed(a_list)` goes through a list from the end to the start.

`update(lr)` just asks every layer to update itself (the `ReLU` layer does nothing).
