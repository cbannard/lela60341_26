# Part 1 · Logical operators and truth tables

In Week 2 you met the **logical operators** `and`, `or` and `not`, which combine `True` and `False` values:

* `a and b` is `True` only when **both** `a` and `b` are `True`
* `a or b` is `True` when **at least one** of them is `True`
* `not a` is `True` when `a` is `False`, and the other way round

# Part 3 · Functions that return several values

In Week 4 you saw that a function can **return several values**, separated by commas. They come back as a **tuple**, which you can **unpack** into separate names, or index like a list:

```python
def minmax(data):
    return min(data), max(data)

low, high = minmax([3, 1, 4])     # unpack: low is 1, high is 4
minmax([3, 1, 4])[1]              # just the second value: 4
```

This is useful when a calculation produces an **intermediate** result that you will need later, as well as the final one: return both, and the caller can keep whichever it needs.

# Part 4 · Building a matrix with nested loops

A matrix is a list of rows (Week 4). To **build** one, use two loops, one inside the other (**nested** loops): the outer loop makes one row each time round, and the inner loop fills in that row, one element at a time. The pattern is always:

```python
matrix = []
for i in range(n_rows):          # one row each time round
    row = []
    for j in range(n_columns):   # one element of the row each time round
        row.append(...)          # the element in row i, column j
    matrix.append(row)
```

For example, here is a multiplication table: the element in row `i`, column `j` is `(i + 1) * (j + 1)`.

````python
table = []
for i in range(3):
    row = []
    for j in range(4):
        row.append((i + 1) * (j + 1))
    table.append(row)
for row in table:
    print(row)
````

# Part 5 · Updating a matrix, and copies

There are two ways to do it, and the difference matters:

* **in place**: change the elements of the existing matrix, `M[i][j] = ...`. Every name that refers to that matrix sees the change.
* **as a new matrix**: build a new matrix with the nested-loop pattern from Part 4, and leave the old one alone.
