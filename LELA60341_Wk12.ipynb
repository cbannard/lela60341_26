# Part 4 · Building a matrix with nested loops

A matrix is a list of rows (Week 4). To **build** one, use two loops, one inside the other (**nested** loops): the outer loop makes one row each time round, and the inner loop fills in that row, one element at a time. The pattern is always:

```python
matrix = []
for i in range(n_rows):          # one row each time round
    row = []
    for j in range(n_columns):   # one element of the row each time round
        row.append(...)          # the element in row i, column j
    matrix.append(row)
```

# Part 5 · A layer with no parameters: ReLU

In Week 8 you built layers as subclasses of `Module`: each has a `forward` method (used through `__call__`) and a `backward` method. A layer doesn't have to have any parameters. A **ReLU layer** just applies `relu` to every number it is given, so it has nothing to learn, but it still needs a `forward` and a `backward`:

* **`forward(X)`** keeps `X` (for the backward pass) and returns a matrix of the same shape with `relu` applied to every number
* **`backward(grad_out)`** passes the gradients back to whatever came before it. Each gradient is multiplied by the **slope** of `relu` at that number (Exercise 3): numbers that were positive pass their gradient back unchanged, and numbers that were cut off to 0 pass back 0, because changing them a little made no difference

Run the next cell first: it sets up the parts from Week 8 (`zeros_like`, `Parameter`, `Module`, `Linear` and `SGD`), with the helper functions `dot`, `mat_vec` and `transpose`.

````python
import math
import random
import matplotlib.pyplot as plt

def dot(a, b):
    # the dot product of two lists of the same length
    total = 0
    for i in range(len(a)):
        total += a[i] * b[i]
    return total

def mat_vec(matrix, vector):
    # the dot product of each row of the matrix with the vector
    return [dot(row, vector) for row in matrix]

def transpose(matrix):
    # the rows of the matrix turned into columns
    return [[row[j] for row in matrix] for j in range(len(matrix[0]))]

def sigmoid(z):
    if z >= 0:
        return 1 / (1 + math.exp(-z))
    else:
        return math.exp(z) / (1 + math.exp(z))      # the same value, without calculating math.exp(-z)

def zeros_like(value):
    # a list or matrix of 0s with the same shape as value
    if isinstance(value[0], list):
        return [[0] * len(row) for row in value]
    else:
        return [0] * len(value)

class Parameter:
    """Numbers that training changes: their values, and their gradients."""
    def __init__(self, value):
        self.value = value
        self.grad = zeros_like(value)

class Module:
    """The parent class of every layer and every model."""
    def __init__(self):
        pass

    def __call__(self, X):
        return self.forward(X)

    def parameters(self):
        params = []
        for value in vars(self).values():
            if isinstance(value, Parameter):
                params.append(value)
            elif isinstance(value, Module):
                params.extend(value.parameters())
        return params

class Linear(Module):
    """A layer that gives n_out scores for each example: one row of weights and one bias per output."""
    def __init__(self, n_in, n_out):
        super().__init__()
        self.weight = Parameter([[random.random() for j in range(n_in)] for c in range(n_out)])
        self.bias = Parameter([random.random() for c in range(n_out)])

    def forward(self, X):
        self._X = X
        out = []
        for x in X:
            z = mat_vec(self.weight.value, x)
            for c in range(len(z)):
                z[c] += self.bias.value[c]
            out.append(z)
        return out

    def backward(self, grad_out):
        W = self.weight.value
        for k in range(len(self._X)):
            for c in range(len(W)):
                for j in range(len(W[c])):
                    self.weight.grad[c][j] += grad_out[k][c] * self._X[k][j]
                self.bias.grad[c] += grad_out[k][c]
        WT = transpose(W)
        return [mat_vec(WT, grad_out[k]) for k in range(len(self._X))]

class SGD:
    """Gradient descent: new value = old value - learning rate * gradient."""
    def __init__(self, parameters, lr):
        self.parameters = parameters
        self.lr = lr

    def zero_grad(self):
        for p in self.parameters:
            p.grad = zeros_like(p.value)

    def step(self):
        for p in self.parameters:
            if isinstance(p.value[0], list):
                for i in range(len(p.value)):
                    for j in range(len(p.value[i])):
                        p.value[i][j] -= self.lr * p.grad[i][j]
            else:
                for i in range(len(p.value)):
                    p.value[i] -= self.lr * p.grad[i]
````

# Part 6 · Layers one after another: `Sequential`

Many models are just layers one after another: the output of each layer is the input of the next. A `Sequential` module holds a list of layers and does three things:

* **`forward(X)`** passes `X` through the layers **in order**: `X = layer(X)` for each layer
* **`backward(grad)`** passes the gradients through the layers **in reverse order**, because the backward pass works from the output back to the input: `grad = layer.backward(grad)` for each layer, last layer first. `reversed(a_list)` goes through a list from the end to the start.
* **`parameters()`** collects the parameters of every layer. It has to **override** `Module.parameters`, because the layers are stored in a list, and `Module.parameters` only looks for parameters and modules stored directly in instance variables.

The constructor takes any number of layers: `Sequential(layer1, layer2, layer3)`. A star before a parameter's name, `*layers`, collects **all** the arguments into a tuple, which we turn into a list.
