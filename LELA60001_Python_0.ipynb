## What is Python?

When we talk about data in computational/corpus linguistics we usually mean text data or audio files. While this data is meaningful to us, to the computer data is always just sequences of numbers. So to a computer the character i is 1101001. Fortunately, we never have to engage with this numeric representation - we can input an i into a computer and it will be converted into the computer's language for us.

A similar situation arises with the instructions that a computer applies to data. The instructions that the computer sees are in a purely numeric format known as machine code. Fortunately, we as users do not need to engage with machine code. Instead, we use what is known as a programming language. This is a much easier to read and write language that we input and which is converted to machine code for us.

The programming language we will employ for this is called Python.

## Variables and datatypes
Data in Python programs is manipulated using variables. A variable is like a box in which you can store information of any kind. You assign data to variables using the assignment operator =. For example you can store the number 5 as the variable `shirt_number` as follows:

```python
shirt_number = 5
```

**Choosing names.** A variable name can contain letters, digits and underscores (`_`), but no spaces, and it can't start with a digit. Names are case-sensitive, so `shirt_number` and `Shirt_number` are different variables. Choose names that say what the variable holds: `shirt_number` is much easier to understand later than `i` or `x`.

Once stored, variables can be manipulated, for example using the standard mathematical operators for addition "+", subtraction "-", multiplication "*" and division "/". For example:

```python
shirt_number + 1 * 3 / 2
```

Did you expect 9? Python follows the usual mathematical rules: `*` and `/` are done before `+` and `-`, so this is `5 + ((1 * 3) / 2)` = 5 + 1.5. Use brackets to choose the order yourself:

```python
(shirt_number + 1) * 3 / 2
```

Data in Python has different types. `shirt_number`, as a whole number, is an integer, or **int**. The results above are **floating point numbers**, or **float**s: numbers with a decimal point. Division with `/` always gives a float, even when the answer is a whole number (`12 / 2` is `6.0`). There are other types of numbers in Python but it is these two you will work with most.

The function `type()` tells you the type of any value:

```python
print(type(shirt_number))
print(type(12 / 2))
```

Variables can also be used to store a letter or a sequence of letters, known as a **string** (type `str`). Strings are written between quotes:

```python
intro = "My name is "
type(intro)
```

Strings can be combined using the "+" symbol, which when applied to strings works as a **concatenation** operator: it joins them together.

```python
full_intro = intro + "Jude Bellingham"
```

Once you have stored information in a variable, you can print it as follows:

```python
print(full_intro)
```

You can also print several values in one go by separating them with commas. `print` puts a space between them, and the values can be of different types:

```python
print(full_intro, "and my shirt number is", shirt_number)
```

### Error messages

Joining with `+` is different: `+` only works when both sides are the same kind of thing. Adding a string and an int makes no sense to Python, so the following gives an **error message**. Run it and have a look.

```python
print(full_intro + " and my shirt number is " + shirt_number)
```

Error messages look alarming, but they are there to help. **Read the last line first**: it names the type of error (here a `TypeError`, meaning a value of the wrong type was used) and describes the problem. The lines above it show where in your code it happened. Everyone who programs sees error messages all the time.

You can avoid this error by explicitly converting the type of a value. `str(x)` turns `x` into a string. (Similarly, `int(x)` and `float(x)` turn `x` into an int or a float, which is useful when you have a number stored as a string, such as `"7"`.)

```python
print(full_intro + " and my shirt number is " + str(shirt_number))
```

There is also a neater way to put values into a string: an **f-string**. Put an `f` before the opening quote, and write any variable inside curly brackets `{}`. Python fills in its value, converting it to a string for you:

```python
print(f"{full_intro} and my shirt number is {shirt_number}")
```

# Lists

So far I have represented sentences as single strings. For most purposes in computational linguistics we don't want to do this - we want instead to represent them in ways that recognize the word boundaries. In order to do this we often represent sentences as lists of words, each of which is represented as a string. A list of strings can be created by putting words (as strings in quotes) inside square brackets, separated by commas, as in the following example:

```python
sentence = ["this", "is", "a", "sentence"]
print(len(sentence))     # for a list, len() gives the number of elements
```

We can print lists of words as a single string when needed using the string method `join`. The string before `.join` sets what goes between the elements of the list. Here we use a space.

```python
print(" ".join(sentence))
print("-".join(sentence))
```

We can also select elements from within the list. The entries in a list are indexed numerically starting with **zero**. So the first element is `sentence[0]` and the last element of this four-element list is `sentence[3]`. You can also count from the end with negative numbers: `sentence[-1]` is the last element and `sentence[-2]` the one before it.

```python
print(sentence[0])
print(sentence[2])
print(sentence[-1])
```

We can also select subsequences of entries (a **slice**) by specifying a range as follows. Notice that the second number in the range isn't included - so `0:2` means from 0 up to the number before 2.

```python
print(" ".join(sentence[0:2]))
```

This allows us to, for example, insert elements in the middle of sentences as follows:

```python
print(" ".join(sentence[0:3]) + " short " + sentence[3])
```

Lists can also be changed. The method `append` adds an element to the end of a list:

```python
words = ["the", "cat"]
words.append("sat")
print(words)
```

# Loading data

Computational linguistics involves handling text data. We saw that we can type in a string of characters, or indeed lists of words, above. However instead of typing in data, we often want to load it from files. We are going to use the English translation of Dostoevsky's novel *Crime and Punishment* from Project Gutenberg: https://www.gutenberg.org/files/2554/2554-0.txt

First we download it to our workspace and save it as the file `crime_and_punishment.txt`. (`import` makes extra Python tools available; we'll say more about this later.)

```python
import urllib.request
urllib.request.urlretrieve("https://www.gutenberg.org/files/2554/2554-0.txt", "crime_and_punishment.txt")
```

Then we read it in to Python. `open()` opens the file, and `.read()` reads the whole of it into one string. `encoding="utf-8"` tells Python how the characters are stored in the file, and the `with` block closes the file again for us when we have finished with it.

```python
with open("crime_and_punishment.txt", encoding="utf-8") as f:
    raw = f.read()

print(len(raw), "characters")
```

We will then extract a single chapter of the novel to work with. The string method `find` gives the position at which a piece of text first occurs. The first chapter starts at the heading "CHAPTER I." and ends where "CHAPTER II." begins, so we can slice out the text between them, just as we sliced lists above:

```python
start = raw.find("CHAPTER I.")
end = raw.find("CHAPTER II.")
print(start, end)

chapter_one = raw[start:end]
print(chapter_one)
```

As you will notice this reads texts in as single strings. As discussed before, we usually want to represent sentences in a way that reflects word boundaries. The simplest way to do it is to split on spaces. As we will see there are all sorts of problems with doing this, but for now we'll ignore that and use the string method `split()`, which splits a string wherever there is whitespace (spaces, tabs or line breaks) and gives a list of the pieces. In order to make sure it deals with full stops as we would like, we will also use the `replace` method to separate them from the ends of words.

```python
chapter_one = chapter_one.replace(".", " .")
chapter_one_tokens = chapter_one.split()
print(chapter_one_tokens[:50])      # the first 50 tokens
```

Now that the chapter is a list of tokens we can start asking corpus-linguistic questions. `len()` counts the tokens, `in` tests whether a word occurs at all, and the list method `count` counts how often it occurs:

```python
print(len(chapter_one_tokens), "tokens")
print("garret" in chapter_one_tokens)
print(chapter_one_tokens.count("garret"))
```

### Regular Expressions

Above we used methods that belong to the string datatype to manipulate text. A much more powerful way of manipulating text is the **regular expression**: a pattern that describes a set of strings. You can find a good overview in section 2.1 of the following chapter:
https://web.stanford.edu/~jurafsky/slp3/2.pdf

Here are the symbols you will need for the exercises below:

| Pattern | Matches | Example | Matches in "the cats sat" |
|---|---|---|---|
| `cat` | the characters c, a, t in that order | `cat` | `cat` |
| `.` | any single character | `s.t` | `sat` |
| `\w` | any "word character": a letter, digit or `_` | `\w\w\w` | `the`, `cat`, `sat` |
| `+` | one or more of the thing before it | `\w+` | `the`, `cats`, `sat` |
| `*` | zero or more of the thing before it | `cats*` | `cats` |
| `[...]` | any one of the characters in the brackets | `[cs]at` | `cat`, `sat` |
| <code>&#124;</code> | either the pattern on the left or the one on the right | <code>cats&#124;sat</code> | `cats`, `sat` |
| `\b` | a word boundary (the start or end of a word) | `\bs\w+` | `sat` |

Once you have looked at that, the following will get you started on regular expressions in Python. First we need to import the regular expressions library (https://docs.python.org/3/library/re.html):

```python
import re
```

### re.findall()
One very useful function in the `re` library is `re.findall()`. This searches for occurrences of a pattern in a given string. It takes a regular expression to search for, between quotes, as its first argument, and a string to search as its second argument. It returns all the matches it finds in a list (an empty list if there are none):

```python
sentence = "I like both dogs and cats"
re.findall(r"cats", sentence)
```

**Always put an `r` before the opening quote of a regular expression**, as in `r"cats"`. This makes it a **raw string**, in which a backslash is just a backslash. Without the `r`, Python gives some backslash combinations a special meaning before `re` ever sees them: `"\b"` means a "backspace" character, so a pattern like `"\bcats"` silently finds nothing.

```python
print(re.findall(r"\bcats", sentence))     # with r: works
print(re.findall("\bcats", sentence))      # without r: finds nothing
```

> ⚠️ If your pattern contains round brackets `( )`, `findall` returns only the part of each match that is inside the brackets. Avoid brackets for now unless you want that.
